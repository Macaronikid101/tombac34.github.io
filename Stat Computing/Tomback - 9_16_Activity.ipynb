```r
library(tidyverse)
df <- read.csv("http://manctsui.github.io/datasets/countries.csv")
head(df, 5)
```

```r
df |>
  filter(Region == "EASTERN EUROPE", Net.migration > 0) |>
  length()
```

```r
df |>
  filter(Region == "EASTERN EUROPE") |>
  arrange(desc(Net.migration)) |>
  head(3)
```

```r
df |>
  filter(Region == "EASTERN EUROPE") |>
  arrange(Net.migration) |>
  head(3)
```

```r
df |>
  filter(Net.migration == NA | Net.migration == 0) |>
  select(Country)
```

```r
df |>
  filter(Area..sq..mi.. > 2000000 | Population > 25000000) |>
  select(Country, Population, Area..sq..mi..)
```

```r
df |>
  filter(Deathrate < Birthrate) |>
  length()

df |>
  filter(is.na(Deathrate) | is.na(Birthrate)) |>
  select(Country)
```

```r
df$Natural_population_change <- df$Birthrate - df$Deathrate
head(df$Natural_population_change, 5)
```

```r
df |>
  arrange(desc(Population)) |>
  head(3)
```

```r
df |>
  filter(Population > 298444215*0.3) |>
  head(3)
```

```r
aggregate(Population ~ Region, data = df, FUN = mean)
```
