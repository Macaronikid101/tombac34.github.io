```r
Countries <- read.csv('http://manctsui.github.io/datasets/countries.csv')
```

```r
head(Countries)
```

```r
length(Countries$Region == "EASTERN EUROPE" & Countries$Net.migration > 0)
```

```r
Countries1 <- Countries[Countries$Region == "EASTERN EUROPE",]
Countries1 <- na.omit(Countries1)
Countries1 <- Countries1[order(Countries1$Net.migration, decreasing = T), ]
head(Countries1, 3)
# Czech Republic, Hungary and Romania
```

```r
Countries1 <- Countries[Countries$Region == "EASTERN EUROPE",]
Countries1 <- na.omit(Countries1)
Countries1 <- Countries1[order(Countries1$Net.migration), ]
head(Countries1, 3)
# Albania, Bulgaria, and Poland
```

```r
Countries$Country[is.na(Countries$Net.migration) | Countries$Net.migration == 0]
```

```r
row_cond <- Countries$Area..sq..mi.. > 2000000 | Countries$Population > 25000000
col_cond <- c("Country", "Population", "Area..sq..mi..")
Countries[row_cond, col_cond]
```

```r
row_cond <- Countries$Birthrate < Countries$Deathrate
length(Countries[row_cond, ]) # 20
length(is.na(Countries$Birthrate) | Countries$Birthrate == 0 | is.na(Countries$Deathrate) | Countries$Deathrate == 0) # 227
```

```r
Countries$Natural_population_change = (Countries$Birthrate - Countries$Deathrate)
Countries$Natural_population_change
```
